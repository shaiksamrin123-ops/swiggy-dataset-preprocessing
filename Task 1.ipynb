```python

# Import pandas to load and work with the dataset
import pandas as pd

# Load the raw Swiggy dataset
file_path = r"C:\Users\ASUS\Downloads\swiggy_file.csv\swiggy_file.csv"
df = pd.read_csv(file_path)

# Display the first five records
display(df.head())

```

```python

# Check the number of rows and columns
print("Dataset Shape:", df.shape)

# Display all column names
print("\nColumn Names:")
print(df.columns.tolist())

# Check data types and non-null counts
print("\nDataset Information:")
df.info()

# Check missing values in each column
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicate records
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python

# Inspect unique formats and sample values in important columns
columns_to_check = [
    "Rating",
    "Number of Ratings",
    "Average Price",
    "Number of Offers",
    "Pure Veg"
]

for column in columns_to_check:
    print(f"\n{column}:")
    print(df[column].value_counts(dropna=False).head(10))

# Display a few records with missing values
print("\nRecords with missing Cuisine:")
display(df[df["Cuisine"].isna()].head())

print("\nRecords with missing Number of Ratings:")
display(df[df["Number of Ratings"].isna()].head())

print("\nRecords with missing Offer Name:")
display(df[df["Offer Name"].isna()].head())
```

```python

# Keep the original dataset unchanged for comparison
df_clean = df.copy()

# Check the shape of the copied dataset
print("Working Dataset Shape:", df_clean.shape)
```

```python

# Replace non-numeric rating labels with missing values
df_clean["Rating"] = df_clean["Rating"].replace(["NEW", "--"], pd.NA)

# Convert the Rating column to numeric format
df_clean["Rating"] = pd.to_numeric(df_clean["Rating"], errors="coerce")

# Check the cleaned ratings
print("Rating Data Type:", df_clean["Rating"].dtype)
print("Missing Ratings:", df_clean["Rating"].isna().sum())

# Display sample rating values
print("\nRating Value Counts:")
print(df_clean["Rating"].value_counts(dropna=False).head(10))
```

```python

# Remove the rupee symbol and text, keeping only the numeric price
df_clean["Average Price"] = df_clean["Average Price"].str.replace(
    r"[^\d.]",
    "",
    regex=True
)

# Convert the cleaned price values to numeric format
df_clean["Average Price"] = pd.to_numeric(
    df_clean["Average Price"],
    errors="coerce"
)

# Check the results
print("Average Price Data Type:", df_clean["Average Price"].dtype)
print("Missing Prices:", df_clean["Average Price"].isna().sum())

# Display sample prices
display(df_clean[["Average Price"]].head(10))
```

```python

# Import regular expressions for identifying numeric rating counts
import re

# Display the most common rating-count formats
print(df_clean["Number of Ratings"].value_counts(dropna=False).head(15))
```

```python

# Convert rating-count labels into numeric values
def clean_rating_count(value):
    # Keep missing values unchanged
    if pd.isna(value):
        return pd.NA

    value = str(value).strip()

    # Exact count is unavailable for this label
    if value == "Too Few Ratings":
        return pd.NA

    # Extract the number and check whether it uses K notation
    match = re.search(
        r"(\d+(?:\.\d+)?)\s*(K)?\+?\s*ratings?",
        value,
        re.IGNORECASE
    )

    if match:
        count = float(match.group(1))

        # Convert K notation into thousands
        if match.group(2):
            count *= 1000

        return int(count)

    # Unrecognized formats are marked as missing
    return pd.NA

# Apply the cleaning function to the rating-count column
df_clean["Number of Ratings"] = (
    df_clean["Number of Ratings"]
    .apply(clean_rating_count)
    .astype("Int64")
)

# Check the converted values
print("Data Type:", df_clean["Number of Ratings"].dtype)
print("Missing Values:", df_clean["Number of Ratings"].isna().sum())
print("\nConverted Values:")
print(df_clean["Number of Ratings"].value_counts(dropna=False).head(10))
```

```python

# Count missing values in every column after numeric conversion
print("Missing Values After Conversion:")
print(df_clean.isnull().sum())

# Calculate the percentage of missing values in each column
missing_percentage = (df_clean.isnull().sum() / len(df_clean)) * 100

print("\nMissing Value Percentage:")
print(missing_percentage.round(2))
```

```python

# Fill missing categorical values with an explicit label
df_clean["Cuisine"] = df_clean["Cuisine"].fillna("Unknown")
df_clean["Offer Name"] = df_clean["Offer Name"].fillna("No Offer Details")
df_clean["Area"] = df_clean["Area"].fillna("Unknown")

# Verify that these columns no longer contain missing values
columns_to_check = ["Cuisine", "Offer Name", "Area"]

print("Missing Values After Filling:")
print(df_clean[columns_to_check].isnull().sum())
```

```python

# Check the range of restaurant ratings
print("Minimum Rating:", df_clean["Rating"].min())
print("Maximum Rating:", df_clean["Rating"].max())

# Count ratings outside the expected 1–5 range
invalid_ratings = df_clean[
    df_clean["Rating"].notna()
    & ~df_clean["Rating"].between(1, 5)
]

print("\nRatings Outside 1–5 Range:", len(invalid_ratings))

# Check whether any prices are zero or negative
print("Zero Prices:", (df_clean["Average Price"] == 0).sum())
print("Negative Prices:", (df_clean["Average Price"] < 0).sum())

# Check the range of restaurant prices
print("Minimum Price:", df_clean["Average Price"].min())
print("Maximum Price:", df_clean["Average Price"].max())
```

```python

# Display the 15 restaurants with the highest listed prices
high_price_restaurants = df_clean.sort_values(
    by="Average Price",
    ascending=False
)

display(
    high_price_restaurants[
        ["Restaurant Name", "Average Price", "Rating", "Cuisine", "Location"]
    ].head(15)
)
```

```python

# Compare the original price text with the cleaned numeric values
price_comparison = pd.DataFrame({
    "Restaurant Name": df.loc[high_price_restaurants.index[:15], "Restaurant Name"],
    "Original Price": df.loc[high_price_restaurants.index[:15], "Average Price"],
    "Cleaned Price": df_clean.loc[high_price_restaurants.index[:15], "Average Price"]
})

display(price_comparison)
```

```python

# Calculate the first and third quartiles of restaurant prices
Q1 = df_clean["Average Price"].quantile(0.25)
Q3 = df_clean["Average Price"].quantile(0.75)

# Calculate the interquartile range
IQR = Q3 - Q1

# Identify unusually low and high prices using the IQR rule
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

price_outliers = df_clean[
    (df_clean["Average Price"] < lower_bound) |
    (df_clean["Average Price"] > upper_bound)
]

# Display the statistical limits and number of outliers
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Price Outliers:", len(price_outliers))

# Show a few high-price outliers
display(
    price_outliers.sort_values("Average Price", ascending=False)[
        ["Restaurant Name", "Average Price", "Location"]
    ].head(10)
)
```

```python
# Count exact duplicate rows in the cleaned dataset
duplicate_count = df_clean.duplicated().sum()

print("Exact Duplicate Rows:", duplicate_count)

# Display a few duplicate records for inspection
display(
    df_clean[df_clean.duplicated(keep=False)]
    .head(10)
)
```

```python
# Store the row count before removing duplicates
rows_before = len(df_clean)

# Remove exact duplicate rows and keep the first occurrence
df_clean = df_clean.drop_duplicates().copy()

# Calculate how many rows were removed
rows_removed = rows_before - len(df_clean)

print("Rows Before Cleaning:", rows_before)
print("Duplicate Rows Removed:", rows_removed)
print("Rows After Cleaning:", len(df_clean))
print("Remaining Duplicate Rows:", df_clean.duplicated().sum())
```

```python
# Check whether missing offer names match restaurants with zero offers
missing_offer = df["Offer Name"].isna()
zero_offers = df["Number of Offers"].eq(0)

print("Missing Offer Names:", missing_offer.sum())
print("Restaurants With Zero Offers:", zero_offers.sum())

print(
    "Missing Offer Name AND Zero Offers:",
    (missing_offer & zero_offers).sum()
)

print(
    "Missing Offer Name BUT Offers Available:",
    (missing_offer & ~zero_offers).sum()
)
```

```python
# Label restaurants with zero offers clearly
df_clean.loc[
    df_clean["Number of Offers"] == 0,
    "Offer Name"
] = "No Offer"

# Verify the updated offer labels
print("No Offer Records:", (df_clean["Offer Name"] == "No Offer").sum())
print("Missing Offer Names:", df_clean["Offer Name"].isna().sum())
```

```python
# Check whether the cleaned dataset has any missing values
print("Missing Values in Each Column:")
print(df_clean.isnull().sum())

# Check the final dataset dimensions
print("\nFinal Dataset Shape:", df_clean.shape)

# Confirm that exact duplicates are removed
print("Remaining Duplicate Rows:", df_clean.duplicated().sum())
```

```python
# Display the data types of all columns
print("Column Data Types:")
print(df_clean.dtypes)
```

```python
# Check the unique values in the Pure Veg column
print("Pure Veg Values:")
print(df_clean["pure_veg"].value_counts(dropna=False))

# Check the most common cuisine categories
print("\nTop 10 Cuisines:")
print(df_clean["cuisine"].value_counts().head(10))

# Check the most common locations
print("\nTop 10 Locations:")
print(df_clean["location"].value_counts().head(10))
```

```python
# Identify rows where rating information is missing
missing_rating = df_clean["rating"].isna()

# Identify rows where the number of ratings is missing
missing_rating_count = df_clean["number_of_ratings"].isna()

# Count missing values and their overlap
print("Missing Rating:", missing_rating.sum())
print("Missing Number of Ratings:", missing_rating_count.sum())

print(
    "Both Missing:",
    (missing_rating & missing_rating_count).sum()
)

print(
    "Only Rating Missing:",
    (missing_rating & ~missing_rating_count).sum()
)

print(
    "Only Number of Ratings Missing:",
    (~missing_rating & missing_rating_count).sum()
)
```

```python
# Select restaurants where both rating fields are missing
missing_rating_records = df_clean[
    df_clean["rating"].isna()
    & df_clean["number_of_ratings"].isna()
]

# Display selected columns for inspection
display(
    missing_rating_records[
        [
            "restaurant_name",
            "rating",
            "number_of_ratings",
            "average_price",
            "pure_veg",
            "location"
        ]
    ].head(10)
)
```

```python
# Calculate the percentage of rows with missing rating information
missing_percentage = (
    df_clean["rating"].isna().sum() / len(df_clean)
) * 100

# Calculate the percentage of rows with available ratings
available_percentage = 100 - missing_percentage

print("Rows with Missing Ratings:", round(missing_percentage, 2), "%")
print("Rows with Available Ratings:", round(available_percentage, 2), "%")
```

```python
# Verify that the cleaned dataset has the expected shape
print("Final Dataset Shape:", df_clean.shape)

# Check for any remaining duplicate records
print("Duplicate Rows:", df_clean.duplicated().sum())

# Check for missing values in every column
print("\nMissing Values by Column:")
print(df_clean.isnull().sum())

# Confirm the numeric columns have valid ranges
print("\nRating Range:", df_clean["rating"].min(), "to", df_clean["rating"].max())
print("Minimum Price:", df_clean["average_price"].min())
print("Minimum Number of Offers:", df_clean["number_of_offers"].min())
```
